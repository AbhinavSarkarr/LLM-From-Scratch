### Step-2: Creating and Assigning Token IDs

```python
#for assigning the token id to the tokens we first need to create a vocabulary from text corpus 
#the vocabulary will consist of all the unique words from the text corpus in a sorted manner with tokens initialized from 0 to all the tokens 

#lets create a funtion for creating vocab 

def create_vocab(text):
    words = convert_to_tokens(text)  #creating tokens from the given text
    unique_words = sorted(list(set(words))) #creating the list of unique words from the tokens in a sorted manner 
    vocab = {unique_words[i]: i for i in range(0, len(unique_words))}
    return vocab

raw_text_vocab = create_vocab(raw_text)
print(raw_text_vocab)
```

```python
#now we got a dictionary which consist of all the unique words from text corpus with their token id mapped 

#we will use this vocab to convert our tokens to token ids which can be understood by the LLM 
#we also need a way to convert this token ids back to words, as the neual network will provide us prediciton in the neumerical format only 

#i am creating a class which will contain both the encoding and the decoding functionality 

class Tokenizer: 
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {id: token for token, id in vocab.items()}

    def convert_to_tokens(self, text):
        splitting = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        words = [word for word in splitting if word.strip()]  #item.strip() will give false if it is a whitespace
        return words 

    def encode(self, text):
        tokens = self.convert_to_tokens(text)
        ids = [self.str_to_int[token] for token in tokens]
        return ids
    
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        # Replace spaces before the specified punctuations
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text

```

```python
tokenizer = Tokenizer(raw_text_vocab)

text = """"It's the last he painted, you know," 
           Mrs. Gisburn said with pardonable pride."""

ids = tokenizer.encode(text)

print(ids)

tokens = tokenizer.decode(ids)
print(tokens)
```

### Stage-3 Adding Special Context Tokens

```python
#the tokenizer that we have created will only work when the words written in the text are present in the vocab used. 
#since there can be alot of words that can be created using character it is difficult to include every word in the vocab 

tokenizer = Tokenizer(raw_text_vocab)

text = """"Hi my name is Abhinav """

ids = tokenizer.encode(text)

print(ids)    #it is giving an error becuase the word "hi" is not present in the vocab that we have created so it was not able to encode it. 
```

```python
#for dealing with such issues we use context tokens like <|unk|>, <|endoftext|> etc
#here i am incorporating these addtional tokens in my current vocab 

#1. unk token will be used to solve the problem of unknown word from the vocab, whenever there is any unknown word this token and its id will be assigned
#2. endoftext will be used to tell my llm that a particular topic or type of text has ended now some different type of text is provided to make it understand better the text

#i am incorporating these two tokens first in my vocab context

print(len(raw_text_vocab))
print(type(raw_text_vocab))


raw_text_vocab['<|unk|>'] = 1130
raw_text_vocab['<|endoftext|>'] = 1131
print(raw_text_vocab)
```

```python
class TokenizerV2: 
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {id: token for token, id in vocab.items()}

    def convert_to_tokens(self, text):
        splitting = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        words = [word for word in splitting if word.strip()]  #item.strip() will give false if it is a whitespace
        return words 

    def encode(self, text):
        tokens = self.convert_to_tokens(text)
        tokens = [
            item if item in self.str_to_int 
            else "<|unk|>" for item in tokens
        ]
        ids = [self.str_to_int[token] for token in tokens]
        return ids
    
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        # Replace spaces before the specified punctuations
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
```

```python
tokenizer = TokenizerV2(raw_text_vocab)

text1 = "Hello, do you like tea?"
text2 = "In the sunlit terraces of the palace."

text = " <|endoftext|> ".join((text1, text2))

print(text)

ids = tokenizer.encode(text)
print(ids)

extracted_text = tokenizer.decode(ids)
print(extracted_text)   #now you can see that the word hello and palace were not present in the vocab we created so it assigned unk token to them 
```

So far, we have discussed tokenization as an essential step in processing text as input to
LLMs. Depending on the LLM, some researchers also consider additional special tokens such
as the following:

[BOS] (beginning of sequence): This token marks the start of a text. It
signifies to the LLM where a piece of content begins.

[EOS] (end of sequence): This token is positioned at the end of a text,
and is especially useful when concatenating multiple unrelated texts,
similar to <|endoftext|>. For instance, when combining two different
Wikipedia articles or books, the [EOS] token indicates where one article
ends and the next one begins.

[PAD] (padding): When training LLMs with batch sizes larger than one,
the batch might contain texts of varying lengths. To ensure all texts have
the same length, the shorter texts are extended or "padded" using the
[PAD] token, up to the length of the longest text in the batch.


However the tokenizer that is used in the GPT do not require any of these tokens like unknown as it uses something called <b>BYTE-PAIR Encoding</b>. <br> Moreover the GPT Tokenizer only uses the <|endoftext|>  token for clarifying the data difference during training
