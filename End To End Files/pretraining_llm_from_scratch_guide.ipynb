## <b>TRAINING THE TOKENIZER</b>

Before we begin training the tokenizer, let’s first discuss the different types of tokenization, along with their advantages and disadvantages.

Tokenization is the process of breaking text into smaller units (tokens), which are then fed into the model. 

There are three main types of tokenization:

1. Word-Level Tokenization
2. Subword-Level Tokenization
3. Character-Level Tokenization

### **1. Word-Level Tokenization**  
In word-level tokenization, each word in the text is treated as an individual token.  

#### **Example:** 
**Sentence:** `"The cat is sleeping"`  
**Tokens:** `["The", "cat", "is", "sleeping"]`  

#### **Pros:**  
**Fewer tokens during inference:** Compared to character-level and subword-level tokenization, the number of tokens is significantly lower, making inference faster.  

#### **Cons:**  
1. **Large vocabulary size:** English alone has over 2 million words, which results in a massive vocabulary that increases memory requirements and slows down processing.  
2. **Out-of-Vocabulary (OOV) problem:** Since the tokenizer only recognizes words seen during training, any new word (e.g., `"pretrained"`) will not be tokenized properly.  
3. **Lack of word structure understanding:** Words with similar roots (e.g., `"dog"` and `"dogs"`) are treated as completely different tokens, reducing efficiency in capturing word relationships.  


### **2. Character-Level Tokenization**  
In this type of tokenization, each character in the text is treated as a separate token.  

#### **Example:**  
**Sentence:** `"The cat"`  
**Tokens:** `["T", "h", "e", " ", "c", "a", "t"]`  

#### **Pros:**  
1. **Small vocabulary size:** Since the tokenizer only needs to recognize characters (e.g., 26 letters in English plus punctuation), the vocabulary is very compact.  
2. **No Out-of-Vocabulary (OOV) problem:** Any new word, no matter how rare, can be tokenized since it is built from characters.  

#### **Cons:**  
1. **Very long token sequences:** Since each word is split into characters, processing even a simple sentence requires many tokens, increasing computational complexity.  
2. **Loss of semantic meaning:** The model has to learn relationships between characters to form meaningful words, making training much harder.  



### **3. Subword-Level Tokenization**  
Subword tokenization strikes a balance between word-level and character-level tokenization. Instead of splitting text strictly by words or characters, it breaks words into frequently occurring subword units. This method is used by tokenizers like **Byte Pair Encoding (BPE), Unigram, and WordPiece**.  

#### **Example:**  
**Sentence:** `"The cats are sleeping"`  
**Tokens:** `["The", "cat", "s", "are", "sleep", "ing"]`  

#### **Pros:**  
1. **Balanced vocabulary size:** Subword tokenization keeps vocabulary manageable while still covering a wide range of words.  
2. **Handles unknown words efficiently:** Rare or new words (e.g., `"pretrained"`) are broken into smaller, meaningful subunits (e.g., `"pre", "train", "ed"`), reducing OOV issues.  
3. **Improved efficiency:** Unlike character-level tokenization, subwords maintain meaning and reduce the number of tokens per sentence.  

#### **Cons:**  
1. **Slightly higher number of tokens compared to word-level tokenization:** Since some words are split into subwords, inference might require more tokens than a purely word-level tokenizer.  
2. **Complexity in implementation:** Training a subword tokenizer requires careful frequency analysis and optimization to find the best subwords.  


## <b>TRAINING THE SUB-WORD TOKENIZER (BYTE-PAIR ALGORITHM)</b>

**Byte Pair Encoding (BPE)** is an advanced tokenization method used in most state-of-the-art (SOTA) language models.
It is designed to efficiently represent words by breaking them into subword units, improving the model's ability to handle
rare or unseen words.
